# Cache Stedin Grid Data for Delft

This notebook fetches all electricity and gas/heat grid data from the Stedin ArcGIS REST API for the entire city of Delft and saves it to pickle files for offline use.

**Output files:**
- `inputs/stedin_cache/delft_gas_grid.pkl` - Gas/heat network data
- `inputs/stedin_cache/delft_lv_elec_grid.pkl` - Low voltage electricity network data  
- `inputs/stedin_cache/delft_transformers.pkl` - MV-LV transformer locations
- `inputs/stedin_cache/cache_metadata.pkl` - Metadata about cache creation

These files can be used in `run_analysis.py` to avoid making Stedin API calls.

````python
import pandas as pd
import geopandas as gpd
import pickle
import os
import sys
from datetime import datetime
import folium
from shapely.geometry import Polygon

# Add parent directory to path to import project functions
sys.path.append('..')

from delft_calliope.functions.process_stedin_grids import fetch_stedin_layer_from_arcgis
````

## Configuration

````python
# Bounding box for entire Delft (lon, lat in WGS84)
# Extended bounds to cover the whole municipality
DELFT_BBOX_COORDS = [
    (4.3388, 51.986),   # Southwest
    (4.37, 51.986),     # Southeast
    (4.37, 52.005),     # Northeast
    (4.3388, 52.005),   # Northwest
    (4.3388, 51.986)    # Close polygon
]

# Stedin ArcGIS FeatureServer URLs and layer IDs
BASE_SERVICE_URL = "https://services-eu1.arcgis.com/IQto421Ac9MzEmFT/arcgis/rest/services/KM_Gasvervangingsdata/FeatureServer"
GAS_LAYER_ID = 1
LV_ELEC_LAYER_ID = 2
TRANSFORMER_LAYER_ID = 6

# Maximum records per request (ArcGIS limit)
MAX_RECORDS = 2000

# Output directory
OUTPUT_DIR = '../delft_calliope/inputs/stedin_cache'
os.makedirs(OUTPUT_DIR, exist_ok=True)

print(f"Delft bounding box: {DELFT_BBOX_COORDS}")
print(f"Output directory: {OUTPUT_DIR}")
print(f"\nStedin ArcGIS Service: {BASE_SERVICE_URL}")
````

## Step 1: Fetch Gas/Heat Network Data

Fetches gas pipeline data from Stedin ArcGIS REST API Layer 1.

````python
print(f"Starting gas network fetch at {datetime.now().strftime('%H:%M:%S')}")
print("Fetching from Stedin ArcGIS REST API...\n")

gas_layer_url = f"{BASE_SERVICE_URL}/{GAS_LAYER_ID}"
stedin_gas_gdf = fetch_stedin_layer_from_arcgis(
    DELFT_BBOX_COORDS, 
    gas_layer_url, 
    "Gas Grid",
    max_records=MAX_RECORDS
)

print(f"\nCompleted at {datetime.now().strftime('%H:%M:%S')}")
print(f"Total gas network features fetched: {len(stedin_gas_gdf)}")
print(f"Coordinate Reference System: {stedin_gas_gdf.crs}")

if len(stedin_gas_gdf) > 0:
    print(f"\nSample columns: {list(stedin_gas_gdf.columns[:10])}")
````

## Step 2: Fetch Low Voltage Electricity Network Data

Fetches LV electricity grid data from Stedin ArcGIS REST API Layer 2.

````python
print(f"Starting LV electricity network fetch at {datetime.now().strftime('%H:%M:%S')}")
print("Fetching from Stedin ArcGIS REST API...\n")

lv_elec_layer_url = f"{BASE_SERVICE_URL}/{LV_ELEC_LAYER_ID}"
stedin_elec_gdf = fetch_stedin_layer_from_arcgis(
    DELFT_BBOX_COORDS, 
    lv_elec_layer_url, 
    "LV Electricity Grid",
    max_records=MAX_RECORDS
)

print(f"\nCompleted at {datetime.now().strftime('%H:%M:%S')}")
print(f"Total LV electricity features fetched: {len(stedin_elec_gdf)}")
print(f"Coordinate Reference System: {stedin_elec_gdf.crs}")

if len(stedin_elec_gdf) > 0:
    print(f"\nSample columns: {list(stedin_elec_gdf.columns[:10])}")
````

## Step 3: Fetch MV-LV Transformer Data

Fetches transformer station locations from Stedin ArcGIS REST API Layer 6.

````python
print(f"Starting MV-LV transformer fetch at {datetime.now().strftime('%H:%M:%S')}")
print("Fetching from Stedin ArcGIS REST API...\n")

transformer_layer_url = f"{BASE_SERVICE_URL}/{TRANSFORMER_LAYER_ID}"
stedin_transformers_gdf = fetch_stedin_layer_from_arcgis(
    DELFT_BBOX_COORDS, 
    transformer_layer_url, 
    "MV-LV Transformers",
    max_records=MAX_RECORDS
)

print(f"\nCompleted at {datetime.now().strftime('%H:%M:%S')}")
print(f"Total transformer features fetched: {len(stedin_transformers_gdf)}")
print(f"Coordinate Reference System: {stedin_transformers_gdf.crs}")

if len(stedin_transformers_gdf) > 0:
    print(f"\nSample columns: {list(stedin_transformers_gdf.columns[:10])}")
````

## Step 4: Verify Data Structure

Check that the data has the expected structure and geometry types.

````python
print("=" * 60)
print("GAS NETWORK DATA STRUCTURE")
print("=" * 60)
print(f"Type: {type(stedin_gas_gdf)}")
print(f"Length: {len(stedin_gas_gdf)}")
print(f"CRS: {stedin_gas_gdf.crs}")
if len(stedin_gas_gdf) > 0:
    print(f"\nColumns: {list(stedin_gas_gdf.columns)}")
    print(f"\nGeometry types: {stedin_gas_gdf.geometry.type.value_counts().to_dict()}")
    print(f"\nFirst row sample:")
    print(stedin_gas_gdf.iloc[0])

print("\n" + "=" * 60)
print("LV ELECTRICITY NETWORK DATA STRUCTURE")
print("=" * 60)
print(f"Type: {type(stedin_elec_gdf)}")
print(f"Length: {len(stedin_elec_gdf)}")
print(f"CRS: {stedin_elec_gdf.crs}")
if len(stedin_elec_gdf) > 0:
    print(f"\nColumns: {list(stedin_elec_gdf.columns)}")
    print(f"\nGeometry types: {stedin_elec_gdf.geometry.type.value_counts().to_dict()}")

print("\n" + "=" * 60)
print("MV-LV TRANSFORMERS DATA STRUCTURE")
print("=" * 60)
print(f"Type: {type(stedin_transformers_gdf)}")
print(f"Length: {len(stedin_transformers_gdf)}")
print(f"CRS: {stedin_transformers_gdf.crs}")
if len(stedin_transformers_gdf) > 0:
    print(f"\nColumns: {list(stedin_transformers_gdf.columns)}")
    print(f"\nGeometry types: {stedin_transformers_gdf.geometry.type.value_counts().to_dict()}")
````

## Step 5: Save to Pickle Files

Save all datasets as GeoDataFrames in pickle format for efficient storage and loading.

````python
# Save gas network
gas_file = os.path.join(OUTPUT_DIR, 'delft_gas_grid.pkl')
with open(gas_file, 'wb') as f:
    pickle.dump(stedin_gas_gdf, f)
print(f" Saved gas network to: {gas_file}")
print(f"  File size: {os.path.getsize(gas_file) / 1024 / 1024:.2f} MB")

# Save LV electricity network
elec_file = os.path.join(OUTPUT_DIR, 'delft_lv_elec_grid.pkl')
with open(elec_file, 'wb') as f:
    pickle.dump(stedin_elec_gdf, f)
print(f" Saved LV electricity network to: {elec_file}")
print(f"  File size: {os.path.getsize(elec_file) / 1024 / 1024:.2f} MB")

# Save transformers
transformer_file = os.path.join(OUTPUT_DIR, 'delft_transformers.pkl')
with open(transformer_file, 'wb') as f:
    pickle.dump(stedin_transformers_gdf, f)
print(f" Saved transformers to: {transformer_file}")
print(f"  File size: {os.path.getsize(transformer_file) / 1024 / 1024:.2f} MB")

# Calculate total size
total_size = (os.path.getsize(gas_file) + os.path.getsize(elec_file) + os.path.getsize(transformer_file)) / 1024 / 1024
print(f"\n Total cache size: {total_size:.2f} MB")
````

## Step 6: Test Loading from Cache

Verify the cached files can be loaded correctly.

````python
# Load from cache
with open(gas_file, 'rb') as f:
    loaded_gas = pickle.load(f)

with open(elec_file, 'rb') as f:
    loaded_elec = pickle.load(f)

with open(transformer_file, 'rb') as f:
    loaded_transformers = pickle.load(f)

# Verify
print("Cache loading test:")
print(f" Loaded {len(loaded_gas)} gas network features")
print(f" Loaded {len(loaded_elec)} LV electricity network features")
print(f" Loaded {len(loaded_transformers)} transformer features")
print(f"\nData types match: {type(loaded_gas) == type(stedin_gas_gdf)} ")
print(f"CRS preserved: Gas={loaded_gas.crs}, Elec={loaded_elec.crs}, Trans={loaded_transformers.crs}")
print("\n Cache is ready to use!")
````

## Step 7: Create Summary Statistics and Metadata

````python
# Calculate statistics
gas_total_length_km = 0
if len(stedin_gas_gdf) > 0:
    gas_gdf_rd = stedin_gas_gdf.to_crs(epsg=28992)
    gas_total_length_km = gas_gdf_rd.geometry.length.sum() / 1000

elec_total_length_km = 0
if len(stedin_elec_gdf) > 0:
    elec_gdf_rd = stedin_elec_gdf.to_crs(epsg=28992)
    elec_total_length_km = elec_gdf_rd.geometry.length.sum() / 1000

# Summary statistics
print("=" * 60)
print("SUMMARY STATISTICS")
print("=" * 60)
print(f"Gas/Heat Network:")
print(f"  Features:        {len(stedin_gas_gdf):,}")
print(f"  Total length:    {gas_total_length_km:.2f} km")
print(f"\nLV Electricity Network:")
print(f"  Features:        {len(stedin_elec_gdf):,}")
print(f"  Total length:    {elec_total_length_km:.2f} km")
print(f"\nMV-LV Transformers:")
print(f"  Count:           {len(stedin_transformers_gdf):,}")
print("=" * 60)

# Save metadata
metadata = {
    'created_at': datetime.now().isoformat(),
    'bounding_box_coords': DELFT_BBOX_COORDS,
    'gas_features': len(stedin_gas_gdf),
    'gas_total_length_km': gas_total_length_km,
    'elec_features': len(stedin_elec_gdf),
    'elec_total_length_km': elec_total_length_km,
    'transformer_count': len(stedin_transformers_gdf),
    'crs': str(stedin_gas_gdf.crs),
    'stedin_service_url': BASE_SERVICE_URL,
    'gas_layer_id': GAS_LAYER_ID,
    'lv_elec_layer_id': LV_ELEC_LAYER_ID,
    'transformer_layer_id': TRANSFORMER_LAYER_ID
}

metadata_file = os.path.join(OUTPUT_DIR, 'cache_metadata.pkl')
with open(metadata_file, 'wb') as f:
    pickle.dump(metadata, f)
    
print(f"\n Saved metadata to: {metadata_file}")
print(f"\nAll files saved successfully!")
````

## Step 8: Optional Quick Visualization

Create an interactive map showing all three networks.

````python
# Calculate map center
center_gdf = stedin_gas_gdf if not stedin_gas_gdf.empty else stedin_elec_gdf
if not center_gdf.empty:
    center_gdf_proj = center_gdf.to_crs(epsg=28992)
    centroids_proj = center_gdf_proj.geometry.centroid
    centroids_wgs = centroids_proj.to_crs(epsg=4326)
    center = [centroids_wgs.y.mean(), centroids_wgs.x.mean()]
else:
    # Fallback to bbox center
    lons = [c[0] for c in DELFT_BBOX_COORDS]
    lats = [c[1] for c in DELFT_BBOX_COORDS]
    center = [sum(lats)/len(lats), sum(lons)/len(lons)]

# Create map
stedin_map = folium.Map(location=center, zoom_start=13, tiles="OpenStreetMap")

# Create feature groups
gas_group = folium.FeatureGroup(name="Gas/Heat Network", show=True)
elec_group = folium.FeatureGroup(name="LV Electricity Network", show=False)
transformer_group = folium.FeatureGroup(name="MV-LV Transformers", show=True)

# Add gas features (sample to avoid browser overload)
if len(stedin_gas_gdf) > 0:
    sample_size = len(stedin_gas_gdf)
    gas_sample = stedin_gas_gdf.sample(n=sample_size) if len(stedin_gas_gdf) > sample_size else stedin_gas_gdf
    for _, row in gas_sample.iterrows():
        folium.GeoJson(
            row.geometry,
            style_function=lambda x: {'color': '#ff5100', 'weight': 2}
        ).add_to(gas_group)
    print(f"Added {len(gas_sample)} gas network features to map (sampled)")

# Add elec features (sample to avoid browser overload)
if len(stedin_elec_gdf) > 0:
    sample_size = len(stedin_elec_gdf)
    elec_sample = stedin_elec_gdf.sample(n=sample_size) if len(stedin_elec_gdf) > sample_size else stedin_elec_gdf
    for _, row in elec_sample.iterrows():
        folium.GeoJson(
            row.geometry,
            style_function=lambda x: {'color': '#3186cc', 'weight': 2}
        ).add_to(elec_group)
    print(f"Added {len(elec_sample)} LV electricity network features to map (sampled)")

# Add transformers
if len(stedin_transformers_gdf) > 0:
    for _, row in stedin_transformers_gdf.iterrows():
        geom = row.geometry
        if geom.geom_type == 'Point':
            location = [geom.y, geom.x]
        else:
            centroid = geom.centroid
            location = [centroid.y, centroid.x]
        
        folium.CircleMarker(
            location=location,
            radius=4,
            color='#8B4513',
            fill=True,
            fillColor='#8B4513',
            fillOpacity=0.7
        ).add_to(transformer_group)
    print(f"Added {len(stedin_transformers_gdf)} transformer features to map")

# Add groups to map
gas_group.add_to(stedin_map)
elec_group.add_to(stedin_map)
transformer_group.add_to(stedin_map)

# Add layer control
folium.LayerControl().add_to(stedin_map)

# Save map
map_file = os.path.join(OUTPUT_DIR, 'delft_stedin_networks_preview.html')
stedin_map.save(map_file)
print(f"\n Saved preview map to: {map_file}")

# Display in notebook
stedin_map
````

## Usage Instructions

To use the cached data in `run_analysis.py`, the workflow has been updated to support offline mode:

```python
# In run_analysis.py config:
CONFIG = {
    'online': False,  # Enable offline mode
    'stedin_cache_path': 'inputs/stedin_cache',  # Path to cache
    # ... other settings
}
```

The cached files will be automatically filtered to the neighborhood bounding box during runtime.
